```python
import pandas as pd
import numpy as np
```

```python
file_path = "../data/Raw/furniture_returns_uncleaned.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
clean_df = df.copy()

print("Original dataset copied for cleaning.")
```

```python
blank_rows = clean_df.isna().all(axis=1).sum()

clean_df = clean_df.dropna(how="all").reset_index(drop=True)

print("Completely blank rows removed:", blank_rows)
print("Dataset shape after removing blank rows:", clean_df.shape)
```

```python
print("Dataset Shape:")
print(clean_df.shape)

print("\nColumn Names:")
print(clean_df.columns.tolist())

print("\nData Types:")
print(clean_df.dtypes)
```

```python
print("Missing Values Before Cleaning:\n")

missing_values = clean_df.isnull().sum()

print(
    missing_values[
        missing_values > 0
    ]
)
```

```python
duplicate_count = clean_df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python
numeric_columns = [
    "Value ₹",
    "Transit Days",
    "SLA hrs",
    "Resale Before ₹",
    "Resale After ₹"
]

print("Numeric Columns:")
print(numeric_columns)
```

```python
for column in numeric_columns:
    
    clean_df[column] = pd.to_numeric(
        clean_df[column],
        errors="coerce"
    )

print("Numeric columns converted successfully!")
```

```python
categorical_columns = [
    "Product",
    "Return Reason",
    "Condition Hint",
    "Severity",
    "Safety",
    "Inspection Outcome"
]

print("Categorical Columns:")
print(categorical_columns)
```

```python
for column in categorical_columns:
    
    clean_df[column] = (
        clean_df[column]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
        .str.lower()
    )

print("Categorical columns standardized successfully!")
```

```python
clean_df["Inspection Outcome"] = clean_df[
    "Inspection Outcome"
].replace({
    "none": "unknown",
    "-": "unknown",
    "n/a": "unknown",
    "nan": "unknown"
})

print("Inspection Outcome standardized successfully!")
```

```python
clean_df["Safety"] = clean_df["Safety"].replace({
    "none": "unknown",
    "-": "unknown",
    "n/a": "unknown",
    
    "minor": "low",
    "moderate": "medium",
    "severe": "high"
})

print("Safety values standardized successfully!")
```

```python
clean_df["Product"] = clean_df["Product"].replace({
    "chiar": "chair"
})

print("Product names standardized successfully!")
```

```python
for column in categorical_columns:
    
    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)
    
    print(clean_df[column].value_counts(dropna=False))
```

```python
for column in numeric_columns:
    
    median_value = clean_df[column].median()
    
    clean_df[column] = clean_df[column].fillna(
        median_value
    )

print("Missing numeric values filled using median!")
```

```python
for column in categorical_columns:
    
    clean_df[column] = clean_df[column].fillna(
        "unknown"
    )

print("Missing categorical values filled with 'unknown'.")
```

```python
before_duplicates = clean_df.shape[0]

clean_df = clean_df.drop_duplicates()

after_duplicates = clean_df.shape[0]

print("Rows before removing duplicates:", before_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

```python
print("Numeric Columns Summary:\n")

print(
    clean_df[numeric_columns]
    .describe()
)
```

```python
for column in numeric_columns:
    
    negative_count = (
        clean_df[column] < 0
    ).sum()
    
    print(
        column,
        "→ Negative values:",
        negative_count
    )
```

```python
for column in numeric_columns:
    
    clean_df.loc[
        clean_df[column] < 0,
        column
    ] = np.nan
    
    clean_df[column] = clean_df[column].fillna(
        clean_df[column].median()
    )

print("Invalid negative values handled successfully!")
```

```python
clean_df["Value Loss ₹"] = (
    clean_df["Resale Before ₹"]
    -
    clean_df["Resale After ₹"]
)

print("Value Loss calculated successfully!")

clean_df[
    [
        "Resale Before ₹",
        "Resale After ₹",
        "Value Loss ₹"
    ]
].head()
```

```python
clean_df["Value Loss %"] = (
    clean_df["Value Loss ₹"]
    /
    clean_df["Resale Before ₹"]
) * 100

clean_df["Value Loss %"] = (
    clean_df["Value Loss %"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

print("Value Loss Percentage calculated successfully!")
```

```python
# Product standardization
clean_df["Product"] = clean_df["Product"].replace({
    "book shelf": "bookshelf",
    "book-shelf": "bookshelf",
    "bedd": "bed",
    "dinning table": "dining table",
    "wardrob": "wardrobe",
    "sofaa": "sofa",
    "cofee table": "coffee table",
    "-": "unknown",
    "none": "unknown"
})


# Return Reason standardization
clean_df["Return Reason"] = clean_df["Return Reason"].replace({
    "none": "unknown",
    "-": "unknown"
})


# Condition Hint standardization
clean_df["Condition Hint"] = clean_df["Condition Hint"].replace({
    "none": "unknown",
    "-": "unknown"
})


# Severity standardization
clean_df["Severity"] = clean_df["Severity"].replace({
    "minor": "low",
    "moderate": "medium",
    "severe": "high",
    "none": "unknown",
    "-": "unknown"
})

print("Final categorical value standardization completed!")
```

```python
print("FINAL CATEGORICAL VALUES\n")

for column in categorical_columns:
    
    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)
    
    print(clean_df[column].value_counts(dropna=False))
```

```python
print("=" * 60)
print("FINAL DATA CLEANING SUMMARY")
print("=" * 60)

print("\nFinal Shape:", clean_df.shape)

print("\nMissing Values:")
print(clean_df.isnull().sum())

print("\nDuplicate Rows:")
print(clean_df.duplicated().sum())

print("\nData Types:")
print(clean_df.dtypes)
```

```python
output_path = "../data/Processed/furniture_returns_cleaned.xlsx"

clean_df.to_excel(
    output_path,
    index=False
)

print("Cleaned dataset saved successfully!")
print("Path:", output_path)
```
