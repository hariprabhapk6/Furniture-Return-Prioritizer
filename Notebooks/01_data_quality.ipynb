```python
missing_data = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_data.sort_values(
    "Missing Count",
    ascending=False
)
```

```python
total_missing = df.isnull().sum().sum()

print("Total missing values:", total_missing)
```

```python
duplicate_rows = df.duplicated().sum()

print("Exact duplicate rows:", duplicate_rows)
```

```python
duplicate_ids = df["Return ID"].duplicated().sum()

print("Duplicate Return IDs:", duplicate_ids)
```

```python
duplicate_records = df[
    df["Return ID"].duplicated(keep=False)
].sort_values("Return ID")

duplicate_records
```

```python
categorical_columns = [
    "Product",
    "Return Reason",
    "Condition Hint",
    "Severity",
    "Safety",
    "Inspection Outcome"
]

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts(dropna=False).head(10))
```

```python
numeric_columns = [
    "Value ₹",
    "Transit Days",
    "SLA hrs",
    "Resale Before ₹",
    "Resale After ₹"
]

for column in numeric_columns:
    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)
    print(df[column].head(20).tolist())
```

```python
for column in numeric_columns:
    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)
    
    print(
        df[column]
        .astype(str)
        .value_counts()
        .head(20)
    )
```

```python
for column in numeric_columns:
    temp = pd.to_numeric(
        df[column],
        errors="coerce"
    )
    
    print(
        column,
        "→ values that cannot be directly converted:",
        temp.isna().sum()
    )
```

```python
for column in numeric_columns:
    
    temp = pd.to_numeric(
        df[column],
        errors="coerce"
    )
    
    negative_count = (temp < 0).sum()
    
    print(
        column,
        "→ negative values:",
        negative_count
    )
```

```python
for column in numeric_columns:
    
    temp = pd.to_numeric(
        df[column],
        errors="coerce"
    )
    
    zero_count = (temp == 0).sum()
    
    print(
        column,
        "→ zero values:",
        zero_count
    )
```

```python
resale_before = pd.to_numeric(
    df["Resale Before ₹"],
    errors="coerce"
)

resale_after = pd.to_numeric(
    df["Resale After ₹"],
    errors="coerce"
)

invalid_resale = df[
    resale_after > resale_before
]

print(
    "Rows where Resale After > Resale Before:",
    len(invalid_resale)
)
```
