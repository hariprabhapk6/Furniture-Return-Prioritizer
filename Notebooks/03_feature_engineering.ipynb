```python
import pandas as pd
import numpy as np
```

```python
file_path = "../data/Processed/furniture_returns_cleaned.xlsx"

df = pd.read_excel(file_path)

print("Cleaned dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
df.head()
```

```python
print("Columns in dataset:")

for column in df.columns:
    print("-", column)
```

```python
important_columns = [
    "Return ID",
    "Product",
    "Value ₹",
    "Transit Days",
    "Condition Hint",
    "Severity",
    "Safety",
    "SLA hrs",
    "Inspection Outcome",
    "Resale Before ₹",
    "Resale After ₹",
    "Value Loss ₹",
    "Value Loss %"
]

df[important_columns].head(10)
```

```python
df["Value Loss ₹"] = (
    df["Resale Before ₹"]
    - df["Resale After ₹"]
)

df["Value Loss ₹"].head()
```

```python
df["Value Loss %"] = np.where(
    df["Resale Before ₹"] > 0,
    (
        df["Value Loss ₹"]
        / df["Resale Before ₹"]
    ) * 100,
    0
)

df["Value Loss %"] = df["Value Loss %"].round(2)

df[
    [
        "Resale Before ₹",
        "Resale After ₹",
        "Value Loss ₹",
        "Value Loss %"
    ]
].head(10)
```

```python
print(df["Severity"].value_counts(dropna=False))
```

```python
severity_map = {
    "low": 1,
    "medium": 2,
    "high": 3,
    "unknown": 1
}

df["Severity Score"] = (
    df["Severity"]
    .map(severity_map)
    .fillna(1)
)

df[
    [
        "Severity",
        "Severity Score"
    ]
].head(10)
```

```python
print(df["Safety"].value_counts(dropna=False))
```

```python
safety_map = {
    "low": 1,
    "medium": 2,
    "high": 3,
    "unknown": 1
}

df["Safety Score"] = (
    df["Safety"]
    .map(safety_map)
    .fillna(1)
)

df[
    [
        "Safety",
        "Safety Score"
    ]
].head(10)
```

```python
print(df["Condition Hint"].value_counts(dropna=False))
```

```python
condition_risk_map = {
    
    # Low Risk
    "no visible damage": 1,
    "minor scratch": 1,
    "scratched surface": 1,
    "color mismatch": 1,
    "wrong fabric": 1,
    "stained fabric": 1,
    "chipped surface": 1,
    "cosmetic damage": 1,
    
    # Medium Risk
    "dent": 2,
    "door misalignment": 2,
    "faulty hinge": 2,
    "loose joints": 2,
    "loose parts": 2,
    "cushion damage": 2,
    "torn fabric": 2,
    "water damage": 2,
    "pest damage": 2,
    
    # High Risk
    "broken frame": 3,
    "broken leg": 3,
    "broken shelf": 3,
    "cracked glass": 3,
    "cracked wood": 3,
    "structural crack": 3,
    "structural damage": 3,
    "severe damage": 3,
    
    # Unknown
    "unknown": 1,
    "Unknown": 1,
    "none": 1,
    "-": 1
}

df["Condition Risk Score"] = (
    df["Condition Hint"]
    .map(condition_risk_map)
    .fillna(1)
)

print("Condition Hint and Risk Score:")

print(
    df[
        [
            "Condition Hint",
            "Condition Risk Score"
        ]
    ].head(10)
)
```

```python
def transit_risk(days):

    if days <= 2:
        return 1

    elif days <= 5:
        return 2

    elif days <= 10:
        return 3

    else:
        return 4


df["Transit Risk Score"] = (
    df["Transit Days"]
    .apply(transit_risk)
)

df[
    [
        "Transit Days",
        "Transit Risk Score"
    ]
].head(10)
```

```python
def assign_value_loss_risk(value_loss_percent):
    
    if value_loss_percent >= 50:
        return 3
    elif value_loss_percent >= 25:
        return 2
    else:
        return 1


df["Value Loss Risk Score"] = (
    df["Value Loss %"]
    .apply(assign_value_loss_risk)
)

print(
    df[
        [
            "Value Loss %",
            "Value Loss Risk Score"
        ]
    ].head(10)
)
```

```python
def assign_sla_urgency(sla_hours):
    
    if sla_hours <= 24:
        return 3
    elif sla_hours <= 48:
        return 2
    else:
        return 1


df["SLA Urgency Score"] = (
    df["SLA hrs"]
    .apply(assign_sla_urgency)
)

print(
    df[
        [
            "SLA hrs",
            "SLA Urgency Score"
        ]
    ].head(10)
)
```

```python
score_columns = [
    "Value Loss ₹",
    "Value Loss %",
    "Value Loss Risk Score",
    "Severity Score",
    "Safety Score",
    "Condition Risk Score",
    "Transit Risk Score",
    "SLA Urgency Score"
]

print("Missing values in engineered features:\n")

print(
    df[score_columns]
    .isnull()
    .sum()
)
```

```python
def sla_urgency(hours):

    if hours <= 24:
        return 4

    elif hours <= 48:
        return 3

    elif hours <= 72:
        return 2

    else:
        return 1


df["SLA Urgency Score"] = (
    df["SLA hrs"]
    .apply(sla_urgency)
)

df[
    [
        "SLA hrs",
        "SLA Urgency Score"
    ]
].head(10)
```

```python
def value_loss_risk(loss_percentage):

    if loss_percentage < 10:
        return 1

    elif loss_percentage < 25:
        return 2

    elif loss_percentage < 50:
        return 3

    else:
        return 4


df["Value Loss Risk Score"] = (
    df["Value Loss %"]
    .apply(value_loss_risk)
)

df[
    [
        "Value Loss %",
        "Value Loss Risk Score"
    ]
].head(10)
```

```python
df["Priority Score"] = (

    df["Value Loss Risk Score"] * 0.30 +

    df["Severity Score"] * 0.20 +

    df["Safety Score"] * 0.20 +

    df["Transit Risk Score"] * 0.15 +

    df["SLA Urgency Score"] * 0.10 +

    df["Condition Risk Score"] * 0.05
)

df["Priority Score"] = df["Priority Score"].round(2)

df[
    [
        "Value Loss Risk Score",
        "Severity Score",
        "Safety Score",
        "Transit Risk Score",
        "SLA Urgency Score",
        "Condition Risk Score",
        "Priority Score"
    ]
].head(10)
```

```python
def assign_priority(score):

    if score >= 3.0:
        return "High"

    elif score >= 2.0:
        return "Medium"

    else:
        return "Low"
```

```python
df["Priority"] = (
    df["Priority Score"]
    .apply(assign_priority)
)

print(df["Priority"].value_counts())
```

```python
def create_evidence(row):

    reasons = []

    if row["Value Loss Risk Score"] >= 3:
        reasons.append("High resale value loss")

    if row["Severity Score"] >= 3:
        reasons.append("High damage severity")

    if row["Safety Score"] >= 3:
        reasons.append("High safety risk")

    if row["Transit Risk Score"] >= 3:
        reasons.append("Long transit time")

    if row["SLA Urgency Score"] >= 3:
        reasons.append("Urgent SLA")

    if row["Condition Risk Score"] >= 3:
        reasons.append("Serious condition issue")

    if not reasons:
        reasons.append("Standard inspection priority")

    return "; ".join(reasons)


df["Priority Evidence"] = (
    df.apply(create_evidence, axis=1)
)

df[
    [
        "Return ID",
        "Priority",
        "Priority Score",
        "Priority Evidence"
    ]
].head(10)
```

```python
high_priority = df[
    df["Priority"] == "High"
]

high_priority[
    [
        "Return ID",
        "Product",
        "Value ₹",
        "Transit Days",
        "Severity",
        "Safety",
        "Value Loss %",
        "Priority Score",
        "Priority Evidence"
    ]
].head(20)
```

```python
priority_summary = pd.DataFrame({
    "Count": df["Priority"].value_counts(),
    "Percentage": (
        df["Priority"]
        .value_counts(normalize=True)
        * 100
    ).round(2)
})

priority_summary
```

```python
feature_columns = [
    "Value Loss ₹",
    "Value Loss %",
    "Severity Score",
    "Safety Score",
    "Condition Risk Score",
    "Transit Risk Score",
    "SLA Urgency Score",
    "Value Loss Risk Score",
    "Priority Score",
    "Priority",
    "Priority Evidence"
]

df[feature_columns].head(10)
```
